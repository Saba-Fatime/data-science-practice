```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Data Source: Public Global Open-Data Archive
url = "https://raw.githubusercontent.com/datasets/covid-19/main/data/countries-aggregated.csv"
df_real = pd.read_csv(url)
# First 5 rows check karaen
df_real.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Plot 1: Top 10 Countries by Total Confirmed Cases
top_countries = df_real.groupby('Country')['Confirmed'].max().nlargest(10).reset_index()
plt.figure(figsize=(10, 5))
sns.barplot(data=top_countries, x='Confirmed', y='Country', palette='viridis')
plt.title('Top 10 Countries by Total Confirmed Cases')
plt.xlabel('Total Confirmed Cases')
plt.ylabel('Country')
plt.show()
# Plot 2: Total Recovered vs Total Deaths (Overall Summary)
total_stats = pd.DataFrame({
    'Metric': ['Confirmed', 'Recovered', 'Deaths'],
    'Total': [df_real['Confirmed'].max(), df_real['Recovered'].max(), df_real['Deaths'].max()]
})
plt.figure(figsize=(6, 4))
sns.barplot(data=total_stats, x='Metric', y='Total', palette='magma')
plt.title('Global Overall Metrics Summary')
plt.ylabel('Count')
plt.show()
```
